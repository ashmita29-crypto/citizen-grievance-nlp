```python
import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Load the raw dataset
df = pd.read_csv("../data/raw/grievances.csv")

# Initialize the lemmatizer
lemmatizer = WordNetLemmatizer()

# Get English stopwords
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    """Clean and normalize a piece of text"""

    # Step 1: Convert to lowercase
    text = text.lower()

    # Step 2: Remove special characters and numbers
    text = re.sub(r'[^a-z\s]', '', text)

    # Step 3: Split into words
    tokens = text.split()

    # Step 4: Remove stopwords and short words
    tokens = [w for w in tokens if w not in stop_words and len(w) > 2]

    # Step 5: Lemmatize words
    tokens = [lemmatizer.lemmatize(w) for w in tokens]

    # Step 6: Join words back into text
    return ' '.join(tokens)

# Apply preprocessing
df['cleaned_text'] = df['complaint_text'].apply(preprocess_text)

# Compare original vs cleaned text
print("ORIGINAL:")
print(df['complaint_text'][0])

print("\nCLEANED:")
print(df['cleaned_text'][0])

# Save cleaned dataset
df.to_csv("../data/processed/grievances_cleaned.csv", index=False)

print(f"\nSaved cleaned dataset: {len(df)} rows")
```
